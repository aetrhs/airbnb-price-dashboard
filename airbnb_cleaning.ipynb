# **Data Cleaning**

Important columns:
- price

- lat
- long

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
# dropping duplicate values first
df.drop_duplicates(inplace=True)
```

```python
df.columns
```

#### **1. price column**

```python
# change price from str to int
df.price = df.price.str.replace('$','')
df.price = df.price.str.replace(',','')

# drop rows with null/NaN values
df.dropna(subset=['price'], inplace=True)
df.price = df.price.astype(int)

print(df.price.dtype)
```

#### **2. room_type column**
There are 4 different types:
- Entire home/apt
- Private room
- Shared room
- Hotel room

```python
print(df['room_type'].value_counts())

# there are 0 null values
print(df['room_type'].isnull().sum())
```

#### **3. lat column (latitude)**

```python
print(df['lat'].value_counts())

# there were 8 null values
print(df['lat'].isnull().sum())

df.dropna(subset=['lat'], inplace=True)
print(df['lat'].isnull().sum())
```

#### **4. long column (longitude)**

```python
print(df['long'].isnull().sum())
df.dropna(subset=['long'], inplace=True)
```

```python
df.isnull().sum()
```

#### **5. neighbourhood_group column**
-> Manhattan, Brooklyn, Queens, Bronx, Staten Island

```python
print(df['neighbourhood_group'].value_counts())
```

```python
# rename the mistyped/typo neighbourhood_group values
df['neighbourhood_group'] = df['neighbourhood_group'].replace({
    'brookln': 'Brooklyn',
    'manhatan': 'Manhattan'
})


print(df['neighbourhood_group'].value_counts())
```

```python
# drop the rows with null neighbourhood_group
df.dropna(subset=['neighbourhood_group'], inplace=True)
print(df['neighbourhood_group'].isnull().sum())
```

#### **6. neighbourhood column**

```python
print(df['neighbourhood'].value_counts())

# there were 15 null values
# drop rows with null neighbourhood values
df.dropna(subset=['neighbourhood'], inplace=True)
print(df['neighbourhood'].isnull().sum())
```

#### **7. construction_year column**

```python
print('Range of construction_year = {} - {}\n' .format(df['construction_year'].min(), df['construction_year'].max()))
print(df['construction_year'].value_counts())

# there were 204 null rows
# fill null values with median, which is 2012
df['construction_year'].fillna(df['construction_year'].median(), inplace=True)
print(df['construction_year'].isnull().sum())
```

```python
df.isnull().sum()
```

#### **8. instant_bookable column** 
Either True or False

```python
print(df['instant_bookable'].value_counts())

# there were 105 null values
# fill the null values with False
df['instant_bookable'].fillna(False, inplace=True)
print(df['instant_bookable'].isnull().sum())
```

#### **9. minimum_nights column**

```python
# clip the min and max values to be from 1-365 only
df['minimum_nights'] = df['minimum_nights'].clip(1, 365)
df['minimum_nights'] = df['minimum_nights'].fillna(df['minimum_nights'].median())
df['minimum_nights'] = df['minimum_nights'].astype('Int64')

print('Range of minimum_nights = {} - {}\n' .format(df['minimum_nights'].min(), df['minimum_nights'].max()))

print(df['minimum_nights'].value_counts())
```

```python
# there were 398 null values
# fill null values with median
print(df['minimum_nights'].median())
```

#### **10. availability_365 column**

```python
print('Range of availability_365 = {} - {}\n' .format(df['availability_365'].min(), df['availability_365'].max()))
# print(df['availability_365'].describe())
# print(df['availability_365'].value_counts())`

# fix the negative values to be 0 and values > 365 to be 365
df['availability_365'].clip(0, 365, inplace=True)

# fill the null values with median
df['availability_365'].fillna(df['availability_365'].median(), inplace=True)
print(df['availability_365'].value_counts())
print(df['availability_365'].isnull().sum())
```

#### **11. cancellation_policy column**

```python
# dealing with cancellation policy 
print(df['cancellation_policy'].value_counts())
df.cancellation_policy.fillna('moderate',inplace=True)
```

#### **12. calculated_host_listings_count column**

```python
print(df['calculated_host_listings_count'].isna().sum())
# dropping rows with null as its only 0.3% of total rows
df.dropna(subset=['calculated_host_listings_count'], inplace=True)
```

#### **13. number_of_reviews column**

```python
print(df['number_of_reviews'].isna().sum())
# dropping rows with null as its only 0.1% of total rows
df.dropna(subset=['number_of_reviews'], inplace=True)
```

#### **14. review_rate_number column**

```python
print(df['review_rate_number'].isna().sum())
# dropping rows with null as its only 0.2% of total rows
df.dropna(subset=['review_rate_number'], inplace=True)
```

```python
# country and country_code columns will be dropped as they contain only 1 value
print(df.country.value_counts())
print(df.country_code.value_counts())
```

```python
# Make sure these columns exist and are in correct type first
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['availability_365'] = pd.to_numeric(df['availability_365'], errors='coerce')
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')  
```

### Derived columns

#### **1. Occupancy Rate**

```python
df['occupancy_rate'] = (365 - df['availability_365']) / 365
```

#### **2. Estimated occupied nights per year**

```python
df['estimated_occupied_nights'] = 365 - df['availability_365']
```

#### **3. Estimated yearly revenue**

```python
df['estimated_revenue'] = df['price'] * df['estimated_occupied_nights']
```
